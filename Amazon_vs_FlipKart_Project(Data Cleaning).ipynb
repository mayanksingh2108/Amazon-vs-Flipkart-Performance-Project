```python
# List of tuples containing (Amazon_df, Flipkart_df, Table_Name)
tables_to_merge = [(a_c_df, f_c_df, 'customers'),
                   (a_p_df, f_p_df, 'products'),
                   (a_cat_df, f_cat_df, 'categories'),
                   (a_o_df, f_o_df, 'orders'),
                   (a_oi_df, f_oi_df, 'order_items'),
                   (a_pay_df, f_pay_df, 'payments'),
                   (a_r_df, f_r_df, 'reviews')]

# Dictionary to hold our final merged dataframes
merged_dfs = {}

for amz, flp, name in tables_to_merge:
    # Add Platform column
    amz['platform'] = 'Amazon'
    flp['platform'] = 'Flipkart'
    
    # Concatenate and store in dictionary
    merged_dfs[name] = pd.concat([amz, flp], ignore_index=True)
    
    print(f"{name} merged successfully! Final shape: {merged_dfs[name].shape}")
```

```python
import os

# 1. Define the name of your new folder
output_folder = "Merged_Ecomm_Data"

# 2. Create the folder if it does not already exist
os.makedirs(output_folder, exist_ok=True)

# 3. Export all merged tables to this specific folder
for name, df in merged_dfs.items():
    # Construct the full file path (e.g., 'Merged_Ecomm_Data/Combined_Customers.csv')
    file_path = os.path.join(output_folder, f"Combined_{name}.csv")
    
    # Save the CSV
    df.to_csv(file_path, index=False)
    print(f"Successfully saved to: {file_path}")
```

```python
from sqlalchemy import create_engine

username = 'root'
password = '1234'
host = 'localhost'         
port = '3306'              
database = 'a_vs_f' 


connection_string = f"mysql+pymysql://{username}:{password}@{host}:{port}/{database}"
engine = create_engine(connection_string)


for table_name, df in merged_dfs.items():
    
    # Prefix the table name so it is clear in your database
    sql_table_name = f"{table_name}"
    
    print(f"Uploading {sql_table_name} to MySQL... Please wait.")
    
    # Write the dataframe to SQL
    # if_exists='replace' will overwrite the table if you run the script multiple times
    # index=False prevents pandas from writing the row numbers as a column
    df.to_sql(name=sql_table_name, con=engine, if_exists='replace', index=False)
    
    print(f"Successfully uploaded {sql_table_name}!")

print("All 7 tables have been successfully pushed to your MySQL database!")
```
